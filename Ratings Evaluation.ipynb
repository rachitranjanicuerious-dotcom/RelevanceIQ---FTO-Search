```python
import numpy as np
import pandas as pd
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

```

```python
df1 = pd.read_csv("1-Patent.csv", encoding="cp1252")

df1 = df1.dropna(subset=["Model Relevance"])

df1 = df1.reset_index(drop=True)


# ============================================================
# 1. MODEL'S REASONING
# ============================================================

# Actual human ratings
y_true = df1["RELEVANCE RATING"]

# Model predictions
y_pred = df1["Model Relevance"]

# Fixed class order
labels = ["H", "M+", "M", "L"]


print("------------------Model's Reasoning-----------------")

print("Classification Report")

print(
    classification_report(
        y_true,
        y_pred,
        labels=labels,
        zero_division=0
    )
)


# Confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=labels
)

print("Confusion Matrix:")
print(cm)


# Display confusion matrix
fig, ax = plt.subplots(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(
    ax=ax,
    values_format="d"
)

plt.title("Relevance Rating Confusion Matrix - 1")
plt.tight_layout()


# SAVE IMAGE
plt.savefig(
    "1_confusion_matrix_model_reasoning.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Close figure
plt.close(fig)


# ============================================================
# 2. FRAMEWORK ONLY
# ============================================================

# Actual human ratings
y_true = df1["RELEVANCE RATING"]

# Framework-only predictions
y_pred = df1["Relevance_Framework_Only"]

# Fixed class order
labels = ["H", "M+", "M", "L"]


print("----------------Framework Only-----------------------")

print("Classification Report")

print(
    classification_report(
        y_true,
        y_pred,
        labels=labels,
        zero_division=0
    )
)


# Confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=labels
)

print("Confusion Matrix:")
print(cm)


# Display confusion matrix
fig, ax = plt.subplots(figsize=(7, 6))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

disp.plot(
    ax=ax,
    values_format="d"
)

plt.title("Relevance Rating Confusion Matrix - 1 (Framework Only)")
plt.tight_layout()


# SAVE IMAGE
plt.savefig(
    "1_confusion_matrix_framework_only.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Close figure
plt.close(fig)
```
